Run the next two cells to import NumPy and create the example input array.

## 2. Your Answer

Complete the `homework()` function below.

- Keep the function name and parameters unchanged.
- Write your code at `# Write your answer here`.
- Do **not** put `import` statements, file downloads, or data-loading code
  inside the function.

The temporary value `None` only keeps the starter cell runnable. Replace it

````python
def homework(a):
    my_result = a[(a % 5 == 0) & (a % 2 == 1)]
    return my_result
````
